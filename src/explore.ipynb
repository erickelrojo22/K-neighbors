```python
import pandas as pd
import json
from sklearn.preprocessing import MultiLabelBinarizer

# Función para manejar errores de JSON
def load_json_safe(json_str, default_value=None):
    try:
        return json.loads(json_str)
    except (TypeError, json.JSONDecodeError):
        return default_value

# Cargar el DataFrame desde el archivo CSV
file_path = "/Users/erickvanscoit/Github/K-neighbors/data/tmdb_5000_moviesk_near.csv"
total_data = pd.read_csv(file_path)

# Verificar columnas disponibles
print("Columnas disponibles en el DataFrame:")
print(total_data.columns)

# Procesar la columna "genres"
if "genres" in total_data.columns:
    total_data["genres"] = total_data["genres"].apply(
        lambda x: [item["name"] for item in load_json_safe(x, default_value=[])] if pd.notna(x) and isinstance(x, str) else None
    )
else:
    print("La columna 'genres' no está disponible en el DataFrame.")

# Procesar la columna "keywords"
if "keywords" in total_data.columns:
    total_data["keywords"] = total_data["keywords"].apply(
        lambda x: [item["name"] for item in load_json_safe(x, default_value=[])] if pd.notna(x) and isinstance(x, str) else None
    )
else:
    print("La columna 'keywords' no está disponible en el DataFrame.")

# Procesar otras columnas como "production_companies" y "production_countries" si están en formato JSON
if "production_companies" in total_data.columns:
    total_data["production_companies"] = total_data["production_companies"].apply(
        lambda x: [item["name"] for item in load_json_safe(x, default_value=[])] if pd.notna(x) and isinstance(x, str) else None
    )

if "production_countries" in total_data.columns:
    total_data["production_countries"] = total_data["production_countries"].apply(
        lambda x: [item["name"] for item in load_json_safe(x, default_value=[])] if pd.notna(x) and isinstance(x, str) else None
    )

# Codificar los géneros con One-Hot Encoding para análisis
if "genres" in total_data.columns:
    mlb = MultiLabelBinarizer()
    genre_encoded = mlb.fit_transform(total_data["genres"].dropna())
    genre_labels = mlb.classes_

    # Crear un DataFrame con los géneros codificados
    genre_df = pd.DataFrame(genre_encoded, columns=genre_labels)
    total_data = pd.concat([total_data.reset_index(drop=True), genre_df.reset_index(drop=True)], axis=1)

    print("Géneros codificados (primeras filas):")
    print(genre_df.head())
else:
    print("No se codificaron los géneros, columna no encontrada.")

# Ver las primeras filas del DataFrame procesado
print("Primeras filas del DataFrame procesado:")
print(total_data.head())
```

```python

total_data["genres"] = total_data["genres"].apply(lambda x: [str(genre) for genre in x])
total_data["keywords"] = total_data["keywords"].apply(lambda x: [str(keyword) for keyword in x])
total_data["production_companies"] = total_data["production_companies"].apply(lambda x: [str(actor) for actor in x])
total_data["production_countries"] = total_data["production_countries"].apply(lambda x: [str(crew_member) for crew_member in x])

total_data["tags"] = total_data["genres"] + total_data["keywords"] + total_data["production_companies"] + total_data["production_countries"]
total_data["tags"] = total_data["tags"].apply(lambda x: ",".join(x).replace(",", " "))

columns_to_drop = ["genres", "keywords", "production_companies", "production_countries"]
existing_columns_to_drop = [col for col in columns_to_drop if col in total_data.columns]
total_data.drop(columns=existing_columns_to_drop, inplace=True)

total_data.iloc[0].tags
```

```python
total_data.to_csv("../data/processed/clean_data.csv", index = False)

conn = sqlite3.connect("../data/movies_database.db")

movies.to_sql("clean_movies_data", conn, if_exists = "replace", index = False)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(total_data["tags"])

model = NearestNeighbors(n_neighbors = 6, algorithm = "brute", metric = "cosine")
model.fit(tfidf_matrix)

def get_movie_recommendations(movie_title):
    movie_index = total_data[total_data["title"] == movie_title].index[0]
    distances, indices = model.kneighbors(tfidf_matrix[movie_index])
    similar_movies = [(total_data["title"][i], distances[0][j]) for j, i in enumerate(indices[0])]
    return similar_movies[1:]

input_movie = "Avatar"
recommendations = get_movie_recommendations(input_movie)
print("Film recommendations '{}'".format(input_movie))
for movie, distance in recommendations:
    print("- Film: {}".format(movie))
```
